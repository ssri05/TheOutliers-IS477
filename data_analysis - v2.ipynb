# Data Cleaning - Belle



## Filter Countries

We filter the three datasets to include five selected countries: the United States, Canada, the United Kingdom, Germany, and Japan. The analysis covers the years 2000–2025, as selected during the initial data analysis.


```python
# Filter datasets to the five selected countries
gdp_clean = gdp[gdp["Country Code"].isin(countries)].copy()
unemp_clean = unemp[unemp["Country Code"].isin(countries)].copy()
infl_clean = infl[infl["Country Code"].isin(countries)].copy()

print("GDP Growth:", gdp_clean.shape)
print("Unemployment:", unemp_clean.shape)
print("Inflation:", infl_clean.shape)
```


## Missing Value Assessment

We examine missing values in each dataset to assess data completeness and identify potential issues before further processing.



```python

# Check missing values in each dataset
datasets = {
    "GDP Growth": gdp_clean,
    "Unemployment": unemp_clean,
    "Inflation": infl_clean
}

for name, df in datasets.items():
    print(f"\n{name}")
    print("Total missing values:", df.isnull().sum().sum())
    print("Missing values by column:")
    print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python

# Identify missing inflation observations
missing_inflation = infl_clean[
    infl_clean[year_columns].isnull().any(axis=1)
]

display(
    missing_inflation[
        ["Country Name", "Country Code", "2025"]
    ]
)
```


### Missing Value Decision

The missing value assessment identified no missing values in the GDP Growth and Unemployment datasets.

One missing value was found in the Inflation dataset for the United States (USA) in 2025.

We decided to retain this missing value as NaN rather than replacing it with zero or an estimated value, since imputation could introduce bias into the economic analysis.



## Duplicate Records Assessment

We check for duplicate rows and duplicate country codes to ensure that each country appears only once in each dataset before reshaping.



```python

# Check duplicate records in each dataset
for name, df in datasets.items():
    print(f"\n{name}")
    print("Duplicate rows:", df.duplicated().sum())
    print(
        "Duplicate country codes:",
        df.duplicated(subset=["Country Code"]).sum()
    )
```


### Duplicate Records Decision

No duplicate rows or duplicate country codes were identified in the GDP Growth, Unemployment, or Inflation datasets.

Therefore, no records were removed during the duplicate checking process.



## Data Standardization

We standardize country codes and convert annual economic indicator values to numeric data types to ensure consistency across the three datasets.


```python

# Standardize country codes and numeric values
for df in [gdp_clean, unemp_clean, infl_clean]:
    df["Country Code"] = (
        df["Country Code"].astype(str).str.strip().str.upper()
    )

    for year in year_columns:
        df[year] = pd.to_numeric(df[year], errors="coerce")

# Check data types
for name, df in datasets.items():
    print(f"\n{name} data types:")
    print(df.dtypes)
```

```python

# Recheck missing values after standardization
for name, df in datasets.items():
    print(f"{name}: {df.isnull().sum().sum()} missing values")
```



## Data Reshaping

The original datasets are organized in wide format, with each year represented as a separate column.

We transform the datasets into long format so that each row represents a country-year observation. This structure facilitates data integration and further analysis.



```python

# Reshape datasets from wide to long format
def reshape_data(df, value_name):
    long_df = df.melt(
        id_vars=["Country Name", "Country Code"],
        value_vars=year_columns,
        var_name="Year",
        value_name=value_name
    )

    long_df["Year"] = long_df["Year"].astype(int)

    return long_df

gdp_long = reshape_data(gdp_clean, "GDP_Growth")
unemp_long = reshape_data(unemp_clean, "Unemployment")
infl_long = reshape_data(infl_clean, "Inflation")

print("GDP Growth:", gdp_long.shape)
print("Unemployment:", unemp_long.shape)
print("Inflation:", infl_long.shape)

display(gdp_long.head(10))
```


## Data Integration and Field Standardization

We merge the three datasets using country codes and years as common identifiers.

Column names are standardized to ensure consistency and improve readability for further analysis.


```python

# Merge GDP Growth and Unemployment
merged_data = gdp_long.merge(
    unemp_long[["Country Code", "Year", "Unemployment"]],
    on=["Country Code", "Year"],
    how="outer",
    validate="one_to_one"
)

# Merge Inflation
merged_data = merged_data.merge(
    infl_long[["Country Code", "Year", "Inflation"]],
    on=["Country Code", "Year"],
    how="outer",
    validate="one_to_one"
)

# Standardize column names
merged_data = merged_data.rename(columns={
    "Country Name": "Country",
    "Country Code": "Country_Code"
})

# Sort by country and year
merged_data = merged_data.sort_values(
    ["Country_Code", "Year"]
).reset_index(drop=True)

print("Merged dataset shape:", merged_data.shape)
display(merged_data.head(10))
```


## Final Data Quality Validation

We validate the merged dataset by checking its dimensions, missing values, duplicate country-year observations, and data types to ensure consistency and completeness.


```python

# Validate the final dataset
print("Dataset shape:", merged_data.shape)

print("\nMissing values:")
print(merged_data.isnull().sum())

print("\nDuplicate country-year records:")
print(
    merged_data.duplicated(
        subset=["Country_Code", "Year"]
    ).sum()
)

print("\nData types:")
print(merged_data.dtypes)

print("\nNumber of countries:",
      merged_data["Country_Code"].nunique())

print("Number of years:",
      merged_data["Year"].nunique())

# Display rows with missing economic indicators
display(
    merged_data[
        merged_data[
            ["GDP_Growth", "Unemployment", "Inflation"]
        ].isnull().any(axis=1)
    ]
)
```


## Numerical Data Quality Assessment

We examine the descriptive statistics and numerical ranges of GDP growth, unemployment, and inflation to identify potential anomalies.

Extreme values are not automatically removed, as they may reflect actual economic conditions rather than data errors.


```python

# Examine numerical distributions
numeric_columns = [
    "GDP_Growth",
    "Unemployment",
    "Inflation"
]

display(merged_data[numeric_columns].describe())

# Check minimum and maximum values
for col in numeric_columns:
    print(f"\n{col}")
    print("Minimum:", merged_data[col].min())
    print("Maximum:", merged_data[col].max())
```


### Numerical Data Quality Findings

The descriptive statistics show that GDP growth ranges from -10.05% to 8.54%, unemployment ranges from 2.35% to 11.19%, and inflation ranges from -1.35% to 8.00%.

All unemployment values fall within the valid percentage range of 0% to 100%. Negative GDP growth and inflation values may reflect actual economic conditions rather than data errors.

No clearly invalid numerical values were identified during this assessment. Therefore, no observations were removed or modified based on their numerical ranges.



## Final Data Cleaning Summary and Export

The data cleaning process included the following steps:

- Filtered the datasets to five selected countries and the years 2000–2025.
- Assessed missing values and retained the missing inflation observation for the United States in 2025 as NaN.
- Checked duplicate records and confirmed that no duplicates were present.
- Standardized country codes and converted economic indicators to numeric data types.
- Reshaped the datasets from wide format to long format.
- Merged GDP Growth, Unemployment, and Inflation using country codes and years.
- Validated the final dataset for missing values, duplicate records, and data type consistency.
- Examined numerical ranges and found no clearly invalid values.

The final cleaned dataset contains 130 country-year observations, covering five countries over 26 years, with three economic indicators.
