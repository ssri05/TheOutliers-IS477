# Data Cleaning - Belle



## Filter Countries

We filter the three datasets to include five selected countries: the United States, Canada, the United Kingdom, Germany, and Japan. The analysis covers the years 2000–2025, as selected during the initial data analysis.


```python
# Filter datasets to the five selected countries
gdp_clean = gdp[gdp["Country Code"].isin(countries)].copy()
unemp_clean = unemp[unemp["Country Code"].isin(countries)].copy()
infl_clean = infl[infl["Country Code"].isin(countries)].copy()

print("GDP Growth:", gdp_clean.shape)
print("Unemployment:", unemp_clean.shape)
print("Inflation:", infl_clean.shape)
```


## Missing Value Assessment

We examine missing values in each dataset to assess data completeness and identify potential issues before further processing.



```python

# Check missing values in each dataset
datasets = {
    "GDP Growth": gdp_clean,
    "Unemployment": unemp_clean,
    "Inflation": infl_clean
}

for name, df in datasets.items():
    print(f"\n{name}")
    print("Total missing values:", df.isnull().sum().sum())
    print("Missing values by column:")
    print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python

# Identify missing inflation observations
missing_inflation = infl_clean[
    infl_clean[year_columns].isnull().any(axis=1)
]

display(
    missing_inflation[
        ["Country Name", "Country Code", "2025"]
    ]
)
```


### Missing Value Decision

The missing value assessment identified no missing values in the GDP Growth and Unemployment datasets.

One missing value was found in the Inflation dataset for the United States (USA) in 2025.

We decided to retain this missing value as NaN rather than replacing it with zero or an estimated value, since imputation could introduce bias into the economic analysis.



## Duplicate Records Assessment

We check for duplicate rows and duplicate country codes to ensure that each country appears only once in each dataset before reshaping.



```python

# Check duplicate records in each dataset
for name, df in datasets.items():
    print(f"\n{name}")
    print("Duplicate rows:", df.duplicated().sum())
    print(
        "Duplicate country codes:",
        df.duplicated(subset=["Country Code"]).sum()
    )
```


### Duplicate Records Decision

No duplicate rows or duplicate country codes were identified in the GDP Growth, Unemployment, or Inflation datasets.

Therefore, no records were removed during the duplicate checking process.



## Data Standardization

We standardize country codes and convert annual economic indicator values to numeric data types to ensure consistency across the three datasets.


```python

# Standardize country codes and numeric values
for df in [gdp_clean, unemp_clean, infl_clean]:
    df["Country Code"] = (
        df["Country Code"].astype(str).str.strip().str.upper()
    )

    for year in year_columns:
        df[year] = pd.to_numeric(df[year], errors="coerce")

# Check data types
for name, df in datasets.items():
    print(f"\n{name} data types:")
    print(df.dtypes)
```

```python

# Recheck missing values after standardization
for name, df in datasets.items():
    print(f"{name}: {df.isnull().sum().sum()} missing values")
```



## Data Reshaping

The original datasets are organized in wide format, with each year represented as a separate column.

We transform the datasets into long format so that each row represents a country-year observation. This structure facilitates data integration and further analysis.



```python

# Reshape datasets from wide to long format
def reshape_data(df, value_name):
    long_df = df.melt(
        id_vars=["Country Name", "Country Code"],
        value_vars=year_columns,
        var_name="Year",
        value_name=value_name
    )

    long_df["Year"] = long_df["Year"].astype(int)

    return long_df

gdp_long = reshape_data(gdp_clean, "GDP_Growth")
unemp_long = reshape_data(unemp_clean, "Unemployment")
infl_long = reshape_data(infl_clean, "Inflation")

print("GDP Growth:", gdp_long.shape)
print("Unemployment:", unemp_long.shape)
print("Inflation:", infl_long.shape)

display(gdp_long.head(10))
```


## Data Integration and Field Standardization

We merge the three datasets using country codes and years as common identifiers.

Column names are standardized to ensure consistency and improve readability for further analysis.


```python

# Merge GDP Growth and Unemployment
merged_data = gdp_long.merge(
    unemp_long[["Country Code", "Year", "Unemployment"]],
    on=["Country Code", "Year"],
    how="outer",
    validate="one_to_one"
)

# Merge Inflation
merged_data = merged_data.merge(
    infl_long[["Country Code", "Year", "Inflation"]],
    on=["Country Code", "Year"],
    how="outer",
    validate="one_to_one"
)

# Standardize column names
merged_data = merged_data.rename(columns={
    "Country Name": "Country",
    "Country Code": "Country_Code"
})

# Sort by country and year
merged_data = merged_data.sort_values(
    ["Country_Code", "Year"]
).reset_index(drop=True)

print("Merged dataset shape:", merged_data.shape)
display(merged_data.head(10))
```


## Final Data Quality Validation

We validate the merged dataset by checking its dimensions, missing values, duplicate country-year observations, and data types to ensure consistency and completeness.


```python

# Validate the final dataset
print("Dataset shape:", merged_data.shape)

print("\nMissing values:")
print(merged_data.isnull().sum())

print("\nDuplicate country-year records:")
print(
    merged_data.duplicated(
        subset=["Country_Code", "Year"]
    ).sum()
)

print("\nData types:")
print(merged_data.dtypes)

print("\nNumber of countries:",
      merged_data["Country_Code"].nunique())

print("Number of years:",
      merged_data["Year"].nunique())

# Display rows with missing economic indicators
display(
    merged_data[
        merged_data[
            ["GDP_Growth", "Unemployment", "Inflation"]
        ].isnull().any(axis=1)
    ]
)
```

## By Data Subject

For each indicator, we will show the descriptive statistics, the distribution of all country-years, and each country's spread.

```python
def show_table(ind):
    t = (desc[desc["indicator"] == ind]
         .drop(columns="indicator").round(2).reset_index(drop=True))
    display(t)

def plot_distribution(col):
    # histograms
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    eda[col].dropna().plot.hist(bins=20, ax=axes[0], edgecolor="white")
    axes[0].set_title(f"Distribution of {labels[col]}")
    axes[0].set_xlabel(labels[col])
    # boxplots
    eda.boxplot(column=col, by="Country_Code", ax=axes[1])
    axes[1].set_title(f"{labels[col]} by country")
    axes[1].set_xlabel("")
    plt.suptitle("")
    plt.tight_layout()
    plt.show()

def leaders(ind):
    sub = desc[desc["indicator"] == ind]
    hi, lo = sub.loc[sub["mean"].idxmax()], sub.loc[sub["mean"].idxmin()]
    vh, vl = sub.loc[sub["std"].idxmax()], sub.loc[sub["std"].idxmin()]
    ext_hi, ext_lo = sub.loc[sub["max"].idxmax()], sub.loc[sub["min"].idxmin()]
    print(f"{labels[ind]}:")
    print(f"- Highest average: {names[hi['country_code']]} ({hi['mean']:.2f}); "
          f"lowest: {names[lo['country_code']]} ({lo['mean']:.2f})")
    print(f"- Most variable (std): {names[vh['country_code']]} ({vh['std']:.2f}); "
          f"steadiest: {names[vl['country_code']]} ({vl['std']:.2f})")
    print(f"- Highest value: {names[ext_hi['country_code']]} {int(ext_hi['max_year'])} ({ext_hi['max']:.2f}); "
          f"lowest value: {names[ext_lo['country_code']]} {int(ext_lo['min_year'])} ({ext_lo['min']:.2f})")
```

```python
rows = []
for code in order:
    g = eda[eda["Country_Code"] == code]
    row = {"Country": names[code]}
    for col in ["Unemployment", "Inflation"]:
        v = g.dropna(subset=[col]).sort_values("Year")
        first, last = v.iloc[0], v.iloc[-1]
        row[f"{col} start ({int(first['Year'])})"] = round(first[col], 2)
        row[f"{col} end"] = round(last[col], 2)
        row[f"{col} end year"] = int(last["Year"])
        row[f"{col} change (pts)"] = round(last[col] - first[col], 2)
    rows.append(row)
display(pd.DataFrame(rows))
```

## Major Economic Events

These tables address the open question in the project plan about how to treat the financial crisis and COVID-19 years. The GDP windows match the team's period definitions in `analyze_indicators.py`. The other windows are our own choices, stated so they can be checked:

- **2008-2010 financial crisis:** GDP trough searched in 2008-2010; unemployment compared with 2007 and its peak searched in 2008-2012.
- **2020-2021 COVID-19:** GDP trough searched in 2020-2021; unemployment compared with 2019 and its peak searched in 2020-2022.
- **Inflation surge:** peak searched in 2021-2025 and compared with the 2015-2019 average.

The data are annual, so short-lived spikes within a year are not visible. "Output back to pre-event level" compounds annual growth rates into an index, so it is an approximation, not official GDP levels.

```python
def gdp_unemp_event(trough_window, base_year, unemp_window):
    rows = []
    for code in order:
        g = eda[eda["Country_Code"] == code].sort_values("Year").set_index("Year")
        level = (1 + g["GDP_Growth"] / 100).cumprod()   # approximate real GDP index

        w = g.loc[trough_window[0]:trough_window[1], "GDP_Growth"]
        t_year, t_val = w.idxmin(), w.min()

        after = g.loc[t_year + 1:, "GDP_Growth"]
        pos = after[after > 0]
        back = level.loc[t_year + 1:]
        back = back[back >= level.loc[base_year]]

        u = g["Unemployment"]
        uw = u.loc[unemp_window[0]:unemp_window[1]]
        rows.append({
            "Country": names[code],
            "GDP trough year": int(t_year),
            "GDP trough (%)": round(t_val, 2),
            "First positive-growth year after": int(pos.index[0]) if len(pos) else "none",
            "Output back to pre-event level": int(back.index[0]) if len(back) else "not yet",
            "Unemp. base": round(u.loc[base_year], 2),
            "Unemp. peak": round(uw.max(), 2),
            "Peak year": int(uw.idxmax()),
            "Rise (pts)": round(uw.max() - u.loc[base_year], 2),
        })
    return pd.DataFrame(rows)
```

### a. 2008-2010 financial crisis

- **GDP:** every country’s trough was 2009, and every one returned to positive growth in 2010. Japan (-5.94) and Germany (-5.54) fell the most, and the U.S. the least (-2.58).

- **Unemployment did not follow GDP depth.** The U.S. had the mildest GDP drop but the largest unemployment rise (4.62 to 9.63, 5.01 points). Germany had the second-deepest GDP drop, yet its unemployment never exceeded its 2007 level in the window (-0.85).

- **Unemployment peaked at different times:** Canada and Germany in 2009, the U.S. and Japan in 2010, and the U.K. in 2012, about three years after its GDP trough.

- **Output recovery** (back to the 2007 level, approximate): the U.S. and Canada in 2010, Germany in 2011, and the U.K. and Japan in 2013.

```python
display(gdp_unemp_event((2008, 2010), 2007, (2008, 2012)))
```

### b. 2020-2021 COVID-19

- **GDP:** every trough was 2020, with positive growth in every country in 2021. The U.K. fell furthest (-10.05) and the U.S. least (-2.08).

- **Unemployment jumped immediately in the U.S. (+4.39) and Canada (+3.97), both peaking in 2020.** Germany (+0.72), Japan (+0.46) and the U.K. (+1.21) saw only small increases. The U.K. peaked in 2021, a year later than the rest.

- **The same pattern as 2009:** the U.S. had the mildest GDP decline but the largest unemployment rise, in both episodes.

- **Output recovery** (back to the 2019 level, approximate): the U.S. and Canada in 2021, and the U.K., Germany and Japan in 2022.

```python
display(gdp_unemp_event((2020, 2021), 2019, (2020, 2022)))
```

### c. Inflation surge

- **Every country rose well above its 2015-2019 average.** The excess was 6.45 points for the U.S., 6.33 for the U.K., 5.73 for Germany, 5.13 for Canada and 2.75 for Japan. They all peaked in 2022, except for Japan in 2023.

- **None had returned to its pre-surge average by the last year.** Canada was closest (2.07 vs. 1.67). The U.K. was still at 3.88 against 1.59. Japan’s last value (3.17 in 2025) is only 0.10 below its peak, so inflation there has barely come down. The U.S. figure is for 2024.

```python
rows = []
for code in order:
    s = eda[eda["Country_Code"] == code].set_index("Year")["Inflation"]
    pre = s.loc[2015:2019].mean()
    w = s.loc[2021:2025].dropna()
    rows.append({
        "Country": names[code],
        "Pre-surge avg (2015-19)": round(pre, 2),
        "Peak (2021-25)": round(w.max(), 2),
        "Peak year": int(w.idxmax()),
        "Excess over pre-surge (pts)": round(w.max() - pre, 2),
        "Last year available": int(w.index[-1]),
        "Last value": round(w.iloc[-1], 2),
    })
display(pd.DataFrame(rows))
```
