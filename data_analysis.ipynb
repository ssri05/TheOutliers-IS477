# Initial Data Analysis - Sri 


```python
import pandas as pd
countries = ["USA", "CAN", "GBR", "DEU", "JPN"]
```

### Checking the shape of each dataset and if the countries we want to analyze exist

```python
gdp = pd.read_csv("data/gdp.csv", skiprows=4)
print("Number of Rows", len(gdp))
print("Number of columns:", len(gdp.columns))
print("Columns:", list(gdp.columns))
```

```python
gdp_countries = gdp[gdp["Country Code"].isin(countries)]
print(list(gdp_countries["Country Code"]))
print(len(gdp_countries)==5)
```

```python
unemp = pd.read_csv("data/unemployment.csv", skiprows=4)
print("Number of Rows", len(unemp))
print("Number of columns:", len(unemp.columns))
print("Columns:", list(unemp.columns))
```

```python
unemp_countries = unemp[unemp["Country Code"].isin(countries)]
print(list(unemp_countries["Country Code"]))
print(len(unemp_countries)==5)
```

```python
infl = pd.read_csv("data/inflation.csv", skiprows=4)
print("Number of Rows", len(infl))
print("Number of columns:", len(infl.columns))
print("Columns:", list(infl.columns))
```

```python
infl_countries = infl[infl["Country Code"].isin(countries)]
print(list(infl_countries["Country Code"]))
print(len(infl_countries)==5)
```

### Keeping only the years 2000-2025

```python
basic_columns = ["Country Name", "Country Code", "Indicator Name","Indicator Code"]
year_columns = [str(year) for year in range(2000,2026)]
gdp = gdp[basic_columns+year_columns]
unemp = unemp[basic_columns+year_columns]
infl = infl[basic_columns+year_columns]
```
